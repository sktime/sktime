```python
from sktime.benchmarking.forecasting import ForecastingBenchmark
from sktime.datasets import load_airline
from sktime.forecasting.timemoe import TimeMoEForecaster
from sktime.forecasting.timesfm2 import TimesFM2Forecaster
from sktime.performance_metrics.forecasting import MeanAbsoluteError
from sktime.split import ExpandingWindowSplitter

benchmark = ForecastingBenchmark(
    backend="loky",
    isolated=True,
    envs_dir=".benchmark_envs",  # two subdirs, one per dependency set
)

benchmark.add_estimator(TimeMoEForecaster("Maple728/TimeMoE-50M"))
benchmark.add_estimator(TimesFM2Forecaster())

cv_splitter = ExpandingWindowSplitter(initial_window=12, step_length=2, fh=6)

scorers = [MeanAbsoluteError()]
dataset_loaders = [load_airline]
for dataset_loader in dataset_loaders:
    benchmark.add_task(
        dataset_loader,
        cv_splitter,
        scorers,
    )

results = benchmark.run("results.csv")
```
